```python
import pandas as pd
```

```python
df = pd.read_excel("Mobile Price Dataset.xlsx")
```

```python
df.head()
```

```python
df.loc[15:30]
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
for col in df.columns:
    if df[col].dtype == "object":
        df[col].fillna(df[col].mode()[0], inplace=True)
    else:
        df[col].fillna(df[col].median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[col] = df[col].clip(lower, upper)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df['price_range'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6)
)
plt.title("Price Range Distribution")
plt.show()
```

```python
df['blue'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6)
)
plt.title("Bluetooth Distribution")
plt.show()
```

```python
sns.countplot(x='dual_sim', data=df)
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(), annot=False, cmap='coolwarm')
plt.show()
```

```python
df.to_excel("Mobile Price Dataset_Cleaned.xlsx", index=False)
```
